```python
# pie chart
```

```python
count = df["dept"].value_counts()
print(count)
```

```python
plt.pie(count, labels = count.index, autopct="%1.1f", explode=[0.1, 0, 0])
plt.axis("equal")
plt.show()
```

```python
#count plot
```

```python
plt.bar(count.index, count, color=["green", "black", "red"])
plt.show()
```

```python
#Bivariate - numerical numerical
```

```python
df["Age"] = [22, 25, 29, 24, 30, 35, 28]
df.head()
```

```python
#Scatter Plot:
```

```python
plt.scatter(df["Age"], df["salary"], color="orange")
plt.show()
```

```python
#line graph
```

```python
sort_age = df.sort_values("Age")
```

```python
plt.plot(sort_age["Age"], df["salary"], color ="red", marker ="*", linewidth = "2")
plt.grid()
plt.show()
```

```python
#Bar chart
```

```python
plt.bar(sort_age["Age"], df["salary"], color ="red")
plt.show()
```

```python
# bivariate numerical - categorical
```

```python
hr_sal = df[df["dept"]=="Hr"]["salary"]
hr_sal
```

```python
it_sal = df[df["dept"]=="IT"]["salary"]
it_sal
```

```python
finace_sal = df[df["dept"]=="Finance"]["salary"]
finace_sal
```

```python
#boxplot
```

```python
plt.boxplot([hr_sal, it_sal, finace_sal], tick_labels=["HR", "IT", "Finance"])
plt.show()
```

```python
#pie chart
```

```python
salary_by_dept = df.groupby("dept").sum()
salary_by_dept
```

```python
plt.pie(salary_by_dept.values, labels=salary_by_dept.index, autopct="%1.2f")
plt.axis("equal")
plt.show()
```

```python
# bar plot

hr_mean = sum(hr_sal)/len(hr_sal)
it_mean = sum(it_sal)/len(it_sal)
fin_mean = sum(finace_sal)/len(finace_sal)


plt.bar(["Hr", "IT", "Finance"], [hr_mean, it_mean, fin_mean], color = ["green", "black", "red"])
plt.show()
```
