```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a: int, b: int):
    """
       it will return the sum of two numbers  

       Args: 
           a: Number one
           B: Number two
    """
    return a + b;
```

```python
@tool
def minus(a: int, b: int):
    """
       it will return the subtractraction of two numbers  

       Args: 
           a: Number one
           B: Number two
    """
    return a - b;
```

```python
add_numbers.invoke({"a": 12, "b": 3})
```

```python
minus.invoke({"a": 12, "b": 3})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent ## link llm and tool
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.7,
    streaming=True
)
```

```python
### we have a tool and llm, now we have to connect to get = agent 
```

```python
agent = create_agent(
    model = llm,
    tools = [add_numbers, minus],
    system_prompt="You are a math teacher, and always use tool for calculation."
)
```

```python
response = agent.invoke({"messages": [{"role": "user", "content": "what is 2+30*10 ?"}]})
```

```python
response["messages"]
```

```python
for res in response["messages"]: ## meko last wala content chaheye
    print(res)
    print("\n")
```
